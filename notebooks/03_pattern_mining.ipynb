# Pattern Mining Analysis
This notebook analyzes the paragraph density and length distribution of HTML elements to prove the validity of our Structural Pattern Mining extraction approach.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Simulating paragraph lengths from a typical DOM tree vs actual article body
np.random.seed(42)
nav_links = np.random.normal(loc=15, scale=5, size=50) # Short nav text
sidebar_widgets = np.random.normal(loc=40, scale=15, size=20) # Sidebar widgets
article_paragraphs = np.random.normal(loc=250, scale=80, size=15) # Dense article text

all_elements = np.concatenate([nav_links, sidebar_widgets, article_paragraphs])
labels = ['Nav']*50 + ['Sidebar']*20 + ['Article']*15

df_dom = pd.DataFrame({'text_length': all_elements, 'element_type': labels})
df_dom = df_dom[df_dom['text_length'] > 0]
```

## Density Visualization
We plot the character length of text nodes. Notice how true article paragraphs are statistically distinct (longer and denser) compared to boilerplate navigation links. This mathematical distinction is what allows our extractor.py to identify the article body without CSS selectors.

```python
plt.figure(figsize=(10, 6))
for element_type in df_dom['element_type'].unique():
    subset = df_dom[df_dom['element_type'] == element_type]
    plt.hist(subset['text_length'], bins=15, alpha=0.6, label=element_type)

plt.axvline(x=100, color='red', linestyle='--', label='Extraction Threshold')
plt.title('DOM Text Node Density Analysis')
plt.xlabel('Character Length')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
